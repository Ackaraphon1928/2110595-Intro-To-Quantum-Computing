```python
#Problem 02
import numpy as np  
sigmay=np.array([[0,-1j],[1j,0]])  
print("Sigma Y:")  
print(sigmay)  
eigenvalues, eigenvectors = np.linalg.eig(sigmay)  
print("Eigenvalues:")  
print(eigenvalues)  
print("Eigenvectors:")  
print(eigenvectors)
```

```python
#Problem 03
import numpy as np  
sigmay=np.array([[0,-1j],[1j,0]])  
print("Sigma Y:")  
print(sigmay)  
print("Sigma Y squared:")  
print(np.matmul(sigmay, sigmay))
```

```python
#Problem 04
import numpy as np  
sigmay=np.array([[1,0],[0,-1]])  
print("Sigma Z:")  
print(sigmay)  
print("Sigma Z squared:")  
print(np.matmul(sigmay, sigmay))
```

```python
#Problem 05
#Show that Sigma Y is Hermitian and unitary.
import numpy as np
sigmay = np.array([[0, -1j], [1j, 0]])
print("Sigma Y:")
print(sigmay)
# Check if Sigma Y is Hermitian
is_hermitian = np.allclose(sigmay, sigmay.conj().T)
print("Is Sigma Y Hermitian?", is_hermitian)
```

```python
#Problem 06
#apply the operator σy to the state ∣0>

sigmay = np.array([[0, -1j], [1j, 0]])
state_0 = np.array([[1], [0]])
result = np.dot(sigmay, state_0)
print("Result of applying Sigma Y to |0>:", result)

#apply the operator σy to the bra ⟨0∣

state_0_dagger = state_0.conj().T
result_dagger = np.dot(state_0_dagger, sigmay)
print("Result of applying Sigma Y to <0|:", result_dagger)
```
