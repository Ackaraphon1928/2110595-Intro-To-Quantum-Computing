```python
import numpy as np

# ============================================================
# Problem 4: Verify Problems 1-3
# ============================================================

# Pauli matrices
sigma_x = np.array([
    [0, 1],
    [1, 0]
], dtype=complex)

sigma_y = np.array([
    [0, -1j],
    [1j, 0]
], dtype=complex)

print("sigma_x =")
print(sigma_x)

print("\nsigma_y =")
print(sigma_y)

```

```python

# ------------------------------------------------------------
# Problem 1: Transformation from sigma_y basis to sigma_x basis
# ------------------------------------------------------------

# Eigenvectors of sigma_x
plus_x  = np.array([[1], [1]], dtype=complex) / np.sqrt(2)
minus_x = np.array([[1], [-1]], dtype=complex) / np.sqrt(2)

# Eigenvectors of sigma_y
plus_y  = np.array([[1], [1j]], dtype=complex) / np.sqrt(2)
minus_y = np.array([[1], [-1j]], dtype=complex) / np.sqrt(2)

# Basis matrices
X = np.hstack((plus_x, minus_x))
Y = np.hstack((plus_y, minus_y))

# Transformation:
# coordinates in y basis -> coordinates in x basis
U_y_to_x = X.conj().T @ Y

print("\nU (sigma_y basis -> sigma_x basis) =")
print(U_y_to_x)

# Check that U is unitary: U†U = I
identity = np.eye(2)

print("\nIs U unitary?")
print(np.allclose(U_y_to_x.conj().T @ U_y_to_x, identity))
```

```python

# ------------------------------------------------------------
# Problem 2: Completeness of sigma_x eigenvectors
# ------------------------------------------------------------

completeness = (
    plus_x @ plus_x.conj().T
    + minus_x @ minus_x.conj().T
)

print("\nCompleteness relation:")
print("|+x><+x| + |-x><-x| =")
print(completeness)

print("\nDoes it equal the identity matrix?")
print(np.allclose(completeness, identity))
```

```python

# ------------------------------------------------------------
# Problem 3: Construct sigma_x from eigenvalues/eigenvectors
# ------------------------------------------------------------

# Eigenvalues:
# |+x> -> +1
# |-x> -> -1

sigma_x_constructed = (
    1 * (plus_x @ plus_x.conj().T)
    - 1 * (minus_x @ minus_x.conj().T)
)

print("\nConstructed sigma_x:")
print(sigma_x_constructed)

print("\nOriginal sigma_x:")
print(sigma_x)

print("\nDo they match?")
print(np.allclose(sigma_x_constructed, sigma_x))
```

```python
# ============================================================
# Problem 5
# f1 ⊗ g2
# g1 ⊗ f2
# ============================================================

fg = np.kron(f, g)
gf = np.kron(g, f)

print("Problem 5")
print("f ⊗ g =")
print(fg)

print("\ng ⊗ f =")
print(gf)

print("\nAre they the same?")
print(np.allclose(fg, gf))

```

```python
# ============================================================
# Problem 6
# f ⊗ g ⊗ f
# ============================================================

fgf = np.kron(np.kron(f, g), f)

print("\n\nProblem 6")
print("f ⊗ g ⊗ f =")
print(fgf)

print("\nDimension:")
print(fgf.shape)
```

```python
# ============================================================
# Problem 7
#
# (f + e) ⊗ g = f ⊗ g + e ⊗ g
#
# g ⊗ (f + e) = g ⊗ f + g ⊗ e
# ============================================================

left1 = np.kron(f + e, g)
right1 = np.kron(f, g) + np.kron(e, g)

left2 = np.kron(g, f + e)
right2 = np.kron(g, f) + np.kron(g, e)

print("\n\nProblem 7")

print("(f + e) ⊗ g =")
print(left1)

print("\nf ⊗ g + e ⊗ g =")
print(right1)

print("\nFirst distributive law verified?")
print(np.allclose(left1, right1))


print("\n\ng ⊗ (f + e) =")
print(left2)

print("\ng ⊗ f + g ⊗ e =")
print(right2)

print("\nSecond distributive law verified?")
print(np.allclose(left2, right2))
```

```python
import numpy as np

# ============================================================
# Problem 11: Verify Problem 10
# ============================================================

# Pauli-X matrix
sigma_x = np.array([
    [0, 1],
    [1, 0]
], dtype=complex)

# Given vectors
f = np.array([
    [0],
    [1]
], dtype=complex)

g = np.array([
    [1 / np.sqrt(2)],
    [1j / np.sqrt(2)]
], dtype=complex)


# ============================================================
# Step 1: Calculate sigma_x |f>
# ============================================================

sigma_f = sigma_x @ f

print("sigma_x |f> =")
print(sigma_f)


# ============================================================
# Step 2: Calculate sigma_x |g>
# ============================================================

sigma_g = sigma_x @ g

print("\nsigma_x |g> =")
print(sigma_g)


# ============================================================
# Step 3:
# (sigma_x |f>) ⊗ (sigma_x |g>)
# ============================================================

left = np.kron(sigma_f, sigma_g)

print("\n(sigma_x |f>) ⊗ (sigma_x |g>) =")
print(left)


# ============================================================
# Step 4:
# |f> ⊗ |g>
# ============================================================

fg = np.kron(f, g)

print("\n|f> ⊗ |g> =")
print(fg)


# ============================================================
# Step 5:
# Construct sigma_x ⊗ sigma_x
# ============================================================

sigma_x_tensor = np.kron(sigma_x, sigma_x)

print("\nsigma_x ⊗ sigma_x =")
print(sigma_x_tensor)


# ============================================================
# Step 6:
# (sigma_x ⊗ sigma_x)(|f> ⊗ |g>)
# ============================================================

right = sigma_x_tensor @ fg

print("\n(sigma_x ⊗ sigma_x)(|f> ⊗ |g>) =")
print(right)


# ============================================================
# Step 7: Compare both results
# ============================================================

print("\nAre the two results equal?")
print(np.allclose(left, right))
```
