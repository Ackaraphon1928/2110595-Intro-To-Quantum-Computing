```python
#Problem 07
import numpy as np

# 1. Define original un-normalized vector in |0>/|1> basis
v1_01 = np.array([1, 2], dtype=complex)

# 2. Define change-of-basis matrix H (Hadamard matrix converting from |0>/|1> to |+>/|->)
H = (1 / np.sqrt(2)) * np.array([[1, 1], 
                                 [1, -1]])

# Project v1 into the |+>/|-> basis: v1_pm = H @ v1_01
v1_pm = H @ v1_01

# 3. Calculate norm and normalize
norm = np.linalg.norm(v1_pm)
v1_pm_normalized = v1_pm / norm

print(f"Un-normalized in |+>/|-> basis: {v1_pm}")
print(f"Norm: {norm:.4f}")
print(f"Normalized in |+>/|-> basis:   {v1_pm_normalized}")
```

```python
#Problem 09
import numpy as np

# Define un-normalized vector in |0>/|1> basis
v1 = np.array([1, 2], dtype=complex)

# Compute norm and normalized state
norm_v1 = np.linalg.norm(v1)
psi = v1 / norm_v1

# Compute measurement probabilities
prob_0 = np.abs(psi[0])**2
prob_1 = np.abs(psi[1])**2

print(f"Norm: sqrt(5) = {norm_v1:.5f}")
print(f"Normalized vector: [1/sqrt(5), 2/sqrt(5)] = {psi}")
print(f"P(|0>): {prob_0:.2f} ({prob_0 * 100:.0f}%)")
print(f"P(|1>): {prob_1:.2f} ({prob_1 * 100:.0f}%)")
```
