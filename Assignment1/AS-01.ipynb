```python
import numpy as np

v1=np.array([[np.sqrt(3)],[1]])  
v2=np.array([[1],[1]])  
inner_result=np.vdot(v1,v2)  
print(inner_result) 
```

```python
v1=np.array([[1j],[1]])  
v2=np.array([[-1j],[1]])  
inner_result=np.vdot(v1,v2)  
print(inner_result) 
```

```python
v1=np.array([[1+2j],[1]])  
v2=np.array([[1-1j],[1]])  
inner_result=np.vdot(v1,v2)  
print(inner_result) 
```
