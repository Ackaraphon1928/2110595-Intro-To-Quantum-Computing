```python
#Problem04-1

import numpy as np

# Pauli matrices
sx = np.array([
    [0, 1],
    [1, 0]
])

sz = np.array([
    [1, 0],
    [0, -1]
])

# Find eigenvalues and eigenvectors of sigma_x
eigenvalues, eigenvectors = np.linalg.eig(sx)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

```python
#Problem04-2
up = np.array([1, 0])
down = np.array([0, 1])

P_up = np.dot(
    up.reshape(2, 1),
    np.transpose(up.reshape(2, 1))
)

P_down = np.dot(
    down.reshape(2, 1),
    np.transpose(down.reshape(2, 1))
)

print("P_up:")
print(P_up)

print("\nP_down:")
print(P_down)
```

```python
#Problem04-3
for i in range(2):

    psi = eigenvectors[:, i]

    # <psi| P_up |psi>
    p_up = np.dot(
        np.transpose(psi),
        np.dot(P_up, psi)
    )

    # <psi| P_down |psi>
    p_down = np.dot(
        np.transpose(psi),
        np.dot(P_down, psi)
    )

    print(f"\nEigenstate {i+1}")
    print("Probability of spin up:  ", p_up)
    print("Probability of spin down:", p_down)
```
